# 1.3 Histograma de frecuencia
 
Es uno de estos gráficos que nos sirven a nosotros, para entender mejor los datos y que está pasando con la información que nos aportan. Es un tipo de gráfico que nos permite ver visualmente las características estadísticas de los datos. Acordaros cuando vimos el cálculo del promedio de la columna goals y como esta no era representativa de la realidad, debido a las características del fútbol. Esta información la obtendremos visualmente gracias al histograma. Porque la forma que tenga un histograma nos dirá hacia qué valores tiende nuestra variable y como es su distribución. 

## 1.3.1 Histograma con matplotlib


```
df.fillna(0, inplace=True)
```

```
plt.hist(df.goals)
plt.show()
```

Como vemos, crear un histograma con matplotlib es tan sencillo como a través de la técnica de punto, accedemos a la función .hist(). Siempre que queramos visualizar nuestro gráfico, al final de la línea de código irá plt.show(), que lo que hace es mostrar el gráfico que hemos creado.
 
Podemos ver un barra muy larga en el valor cero, ya que hay muchos jugadores en la liga que marcan entre 0 y 2 goles. Con estos gráficos podemos ver sobre qué valores se concentran la gran mayoría de los datos. Se puede ver como hay muy pocos jugadores con muchos goles y muchos jugadores con pocos goles.

###1.3.1.1 Paramatero bins
 
Como hemos visto en la imagen, se nos muestra sobre qué rangos de valores se acumulan los datos. Estos rangos, que son la anchura de cada barra del histograma se pueden especificar a través del parámetro bins, si no lo le especificamos ningún valor, la función plt.hist() realizará los rangos con el valor predeterminado del parámetro bins. Para entenderlo veamos un histograma con bins=5.

```
plt.hist(df.goals, bins=5)
plt.show()
```

Hemos dividido la cantidad de goles en rangos de 5, por tanto la primera  barra contiene los jugadores que van desde 0 hasta 5 goles. Veamos qué pasa si utilizamos un bins=25


```
plt.hist(df.goals, bins=25)
plt.show()
```

Vemos cómo obtenemos más barras, pero de rangos más estrechos. Si nos fijamos, el valor más alto en goles es 25, por tanto el rango de valores de la columna goals va desde 0 a 25, al utilizar un bins=25, estamos dividiendo ese rango entre 25, por tanto estamos obteniendo una barra para cada gol. La primera barra agrupa a los jugadores  que tienen 0 goles y la segunda barra a los jugadores con 1 gol y así hasta llegar al valor 25. Dependiendo de cómo quieras ver la información utilizarás diferentes valores para el parámetro bins.
 
Esta es una manera de obtener la información visual que nos aportaba la funcion .value_counts()


```
df.goals.value_counts()
```

###1.3.1.2 Parametro density
 
Hasta ahora el histograma nos mostraba la cantidad de jugadores incluidos en un rango de valores. Con el parámetro density=True, podemos saber la cantidad de jugadores en forma de porcentaje dentro de cada rango de valores. 

```
plt.hist(df.goals, density=True, bins=25)
plt.show()
```

Ahora en vez de ver la cantidad de jugadores de manera numérica, podemos ver como el 50% de los jugadores no meten goles. 
 
Esta misma información, también la podíamos obtener con el parámetro normalize=True de .value_counts() donde nos dirá el resultado en forma de porcentaje.

```
df.goals.value_counts(normalize=True)
```

### 1.3.1.3 Añadir titulo y nombre a los ejes
Matplotlib nos permite añadir titulo al grafico y a los ejes. Esto es importante para que cuando miremos un gráfico, saber qué información se está mostrando. Podemos añadir título y nombre a los ejes con los diferentes métodos.
- Añadir título: plt.title()
- Añadir nombre al eje x: plt.xlabel()
- Añadir nombre al eje y: plt.ylabel()


```
plt.hist(df.goals, density=True, bins=25)
plt.title('Distribución del total de goles')
plt.xlabel('Goles')
plt.ylabel('Proporción')
plt.show()
```

Ahora tenemos un gráfico en su forma correcta. Con el título y el nombre de los ejes para saber qué información está mostrando. 
 
 
### 1.3.1.4. Cambiar el color 
Por defecto el color del histograma es azul, pero matplotlib en su graficos tiene el parámetro color, al cual podremos especificar el color que queramos darle al histograma. 

```
plt.hist(df.goals, density=True, bins=25, color='green')
plt.title('Distribución del total de goles')
plt.xlabel('Goles')
plt.ylabel('Proporción')
plt.show()
```

Este parámetro color lo encontraremos en todos los gráficos de matplotlib.

## 1.3.2. Histograma con seaborn

Seaborn es otra librería de python que nos permite crear visualizaciones. Cada librería tiene sus características, a medida que las vayas utilizando, irás sabiendo cuál utilizar. De momento aprenderás a cómo realizar el mismo gráfico en las diferentes librerías. Para crear un histograma con seaborn utilizamos la función sns.displot()


```
sns.displot(df.goals, bins=25)
plt.title('Distribución del total de pases')
plt.xlabel('Pases')
plt.ylabel('Frecuencia')
plt.show()
```

Como vemos, podemos utilizar la librería seaborn y matplotlib conjuntamente. Estamos creando el gráfico con seaborn, pero  añadiendo título, nombre a los ejes y mostrando el gráfico con matplotlib. Las dos librerías funcionan perfectamente en conjunto. Siempre recuerda que al finalizar un gráfico, tanto con seaborn como con matplotlib, utilizaremos plt.show() para mostrar el resultado final.
 
Cada librería tiene sus peculiaridades. Con seaborn obtenemos un histograma pero con una estética diferente a matplotlib. Eres libre de elegir la que más te guste. 


### 1.3.2.1 Parametro kind
 
Dependiendo del valor que le pongamos al parámetro kind, nos permite crear un histograma con un aspecto visual diferente. Tenemos tres posibles valores a pasar al parámetro kind 
 
- ``'hist'``:
- ``'kde'``:
- ``'ecdf'``:
 
De manera predeterminada el parámetro kind tiene el valor 'hist'. Veamos qué tipos de histogramas obtenemos con los demás valores. 

#### 1.3.2.1.1 kind='kde' 
 
Cuando el valor de kind sea 'kde', obtendremos un histograma suavizado que se conoce como gráfico de densidad.

```
sns.displot(df, x='goals', kind='kde')
plt.show()
```

#### 1.3.2.1.2 kind='ecdf'
Con kind='ecdf', estamos obteniendo la empirical cumulative distribution function, que nos permite observar el porcentaje de acumulación de un valor.  

```
sns.displot(df.goals, kind='ecdf')
plt.show()
```

Lo que nos dice este tipo de gráfico, es la cantidad de jugadores en forma de porcentaje, que hay un un valor determinado de goles. Es parecido a los percentiles, en este caso vemos como el 80 % de los jugadores meten 5 o menos goles.  

###1.3.2.2. Parámetro hue

El parámetro hue nos permite mostrar diferentes histogramas  en función de diferentes categorías, como las posiciones. Tendremos un histograma de valores para cada posición. Podemos ver cómo se distribuyen los goles para cada tipo de posición. Como el portero no participa en los goles, filtramos los datos para las demás posiciones.

```
sns.displot(df[df.position != 'Goalkeeper'], x='goals', kind='kde', hue='position')
plt.title('Distribución de los goles para las diferentes categorías')
plt.show()
```

Ahora en un solo gráfico podemos observar cómo se distribuye la cantidad de goles para las diferente categorías. Lo ancho o estrecho de la curva nos dirá como de concentrados se encuentran los valores. Si nos fijamos los delanteros tienen una curva muy ancha, por lo que los valores de goles están más distribuidos. En cambio tanto para defensa como mediocentro se centran más los valores entorno a una cantidad
 
También podemos ver como cambia la distribución acumulada para cada posición utilizando kind='ecdf' y hue='position'


```
sns.displot(df[df.position != 'Goalkeeper'], x='goals', kind='ecdf', hue='position')
plt.title('Distribución acumulada de los goles para las diferentes categorías')
plt.show()
```

### 1.3.2.2. Parámetro col 
Con el parámetro col podremos mostrar las distribuciones de una variables numérica en función de una variable categórica como hacíamos con hue. Pero esta vez por separado. Se mostrará un gráfico para cada posición.

```
sns.displot(df[df.position != 'Goalkeeper'], x='goals', kind='kde', col='position',)
plt.show()
```

De esta manera podemos ver mejor cómo se distribuyen los datos para cada categoría y sobre qué valores se centran.

## 1.3.3 Histograma con plotly 
 
Plotly es otra librería de visualización muy avanzada en python. A diferencia de seaborn y matplotlib, los gráficos de plotly son interactivos. Cuando pasemos el ratón por encima del gráfico nos mostrará información de los datos que contiene. Para realmente apreciar la interactividad que nos proporciona plotly con los gráficos, ve al video de la clase donde te la enseño, ya que en las siguientes imágenes no se apreciara. 

```
px.histogram(nba, x='PTS', title='Points per Game',
             labels={'PTS':'Puntos por partido'})
```

Con Plotly ya no añadimos titulo ni mostramos el gráfico como hacíamos con matplotlib y seaborn. Estos valores ahora se añaden como parámetro dentro de la función px.histogram() de plotly.
 
- title: Añadimos el título
- labels: Añadimos el nombre de los ejes en formato de diccionario,siendo la llave la columna y el valor el nombre que queremos que se vea en el eje. 

### 1.3.3.1 Parametro facet_col
El parámetro facet_col realiza la misma función que hacía el parámetro col con seaborn. Hará un histograma para las diferentes categorías de una columna categórica. En este  caso las posiciones en baloncesto

```
px.histogram(nba, x='PTS', facet_col='Pos', title='Points per position')
```

De esta manera obtenemos un histograma para cada posición. 
 
### 1.3.3.2 La funcion .update_layout()
Hemos visto como por los parámetros de px.histogram() podíamos añadir título y nombre a los ejes, pero una vez creado un gráfico con los títulos, podemos realizar modificaciones como el color, el tipo de letra y el tamaño de la letra. La única diferencia es que cuando creamos el gráfico lo guardamos en una variable.


```
fig = px.histogram(nba, x='PTS', title='Points per Game',
                   labels={'PTS':'Puntos por partido'})
```

Después es esa variable que contiene un gráfico, le aplicamos a través de punto la función .update_layout() que nos permitirá hacer actualizaciones al gráfico a través de sus parámetros

- title_font_family: El tipo de tipografía de la letra
- title_font_color: El color de la letra
- title_font_size: El tamaño que queramos darle 

```
fig.update_layout(title_font_family='Courier New',
                  title_font_color= 'red',
                  title_font_size=25)
```

Como vemos todas estas modificaciones se las hemos aplicado al título del gráfico.

#1.4. Gráfico de barras
Acordaros, que cuando evaluamos una columna categórica, utilizamos la función .value_counts() para ver los diferentes valores, dentro de una columna categórica y la de veces que aparece cada valor. 


```
df.team.value_counts()
```

Tenemos un método en seaborn que nos permite crear este tipo de visualizaciones para estos casos. 
 
##1.4.1 El método .countplot()
Con .countplot() podremos graficar una columnas categórica y veremos visualmente la frecuencia de aparición de cada valor de la columna.


```
plt.figure(figsize=(12,8))
sns.countplot(x=df.team)
plt.show()
```

Obtenemos de manera visual la cantidad de veces que se repite en la columna team cada equipo, en este caso nos está dando la información de la cantidad de jugadores de cada equipo. 
 
### 1.4.1.1 El metodo .xticks()
Ya vimos que matplotlib y seaborn las podemos usar conjuntamente. Si nos fijamos en el gráfico, vemos como en el eje x, no se puede ver los nombres de los equipos de cada barra, por la cantidad de valores que tenemos. Para estos casos tenemos el método .xticks(), que nos permite hacer modificaciones a los valores del eje x. El método .xticks() tiene el  parámetro  rotation, con el cual podremos rotar los valores del eje x. Si le damos el valor de 90, el nombre de los equipos pasará de estar en horizontal a vertical. De esta manera podremos ver bien los nombres de los equipos. 


```
plt.figure(figsize=(12,8))
sns.countplot(x=df.team)
plt.xticks(rotation=90)
plt.show()
```

### 1.4.1.2. Cambiar de color
Para la librería seaborn también tenemos el parámetro color para darle al gráfico el color que queramos.


```
plt.figure(figsize=(12,8))
sns.countplot(x=df.team, color='blue')
plt.xticks(rotation=90)
plt.show()
```

## 1.4.2. El parámetro hue
Si nos empezamos a dar cuenta, todos los métodos de la librería seaborn contienen el parámetro hue. El parámetro hue, nos va a permitir mostrar más información en el gráfico, gracias a poder añadir la información de otra columna categórica. Ahora estamos viendo la cantidad de jugadores en cada equipo, gracias a hue podemos añadir los valores de la columna 'position' y así mostrar las cantidad de posiciones en cada equipo. 


```
plt.figure(figsize=(12,8))
sns.countplot(x='team', hue='position', data=df)
plt.xticks(rotation=90)
plt.show()
```

Ahora de un vistazo podemos ver la cantidad de diferentes posiciones que contiene cada equipo. Podemos ver como el Getafe, Español y Levante tienen más defensas que cualquier otra posición.
 
 
## 1.4.3. Convertir el gráfico vertical a horizontal 
Ya vimos que al crear el gráfico con .countplot() teníamos el parámetro x, donde le especificamos los valores a graficar. Si en vez de usar el parámetro x utilizamos el parámetro y, nos mostrará las barras de manera horizontal.

```
plt.figure(figsize=(12,8))
sns.countplot(y='team', hue='position',data=df)
plt.xticks(rotation=90)
plt.show()
```

Utilizamos el parámetro y porque ahora los diferentes equipos aparecen en el eje y. Si nos fijamos ya no nos hace falta utilizar .xticks() para rotar 90 grados los nombres de los equipos. De esta manera ya no se estorban entre ellos.

Dependiendo de lo que estemos buscando representar en los gráficos, lo haremos de una manera o de otra. Para ambos casos existen diferentes tipos de gráficos para abordar las diferentes situaciones.

# 1.4 Bar plot
Cuando utilizamos la función .groupby() realizamos agrupaciones en función de una columna categórica, para después analizar otra columna numérica en función de las diferentes agrupaciones de la columna categórica. El gráfico de barras lo que haces es representar visualmente esta agrupación. Por tanto para el gráfico de barras tendremos el eje x y el eje y. En el eje x ira la columna categórica y en el eje y la columna numérica que queramos analizar en función de las categorías. 


## 1.4.1. Matplotlib
La manera de crear un gráfico de barras es con plt.bar() por parámetro solo necesitaremos indicarle las columnas. Podemos ver visualmente la cantidad de goles de cada equipo.

```
plt.bar(df.team, df.goals)
plt.title('Goals by team')
plt.xlabel('Teams')
plt.ylabel('Goals')
plt.xticks(rotation=90)
plt.show()
```

### 1.4.1.1. plt.barh()
Podemos mostrar la misma información de manera horizontal con el método plt.barh(), donde cambiaremos el valor de los ejes, la columna categórica pasará al eje y y el valor de la columnas numéricas al eje x.

```
plt.barh(df.team, df.goals)
plt.title('Goals by team')
plt.ylabel('Teams')
plt.xlabel('Goals')
plt.show()
```

### 1.4.1.2 Hacer el grafico mas grande
Hasta ahora se nos ha mostrado un gráfico en un tamaño predeterminado, pero con plt.figure() podremos darle el tamaño que queramos al gráfico, ya que plt.figure() tiene el parámetro figsize, entre paréntesis tendremos que indicar el tamaño siendo el primer valor la anchura y el segundo la altura.

```
plt.figure(figsize=(10,8))
plt.barh(df.team, df.goals)
plt.title('Goals by team',fontsize=20)
plt.ylabel('Teams', fontsize=15)
plt.xlabel('Goals', fontsize=15)
plt.show()
```

Ahora tenemos un gráfico más grande, yo recomiendo siempre utilizar plt.figure() para darle un tamaño más grande a los gráficos.
## 1.4.2 Seaborn 
 
Con la librería seaborn podemos hacer un gráfico de barras con sns.barplot(). Vamos a realizar un gráfico de barras de los goles en función de las posiciones, filtrando el dataframe para no incluir a los porteros. 

```
sns.barplot(x='position', y='goals',data=df[df.position != 'Goalkeeper'])
plt.title('Average goals by position')
plt.show()
```

Si nos fijamos vemos que este gráfico de barras no muestra el total, si no el promedio de los valores. Tiene un parámetro, que se llama estimator que nos permite mostrar otro tipo de información a parte del total. De forma predeterminada si no se lo especificamos utilizará el promedio. Si queremos ver el total tendremos que utilizar  estimator=sum . 

```
sns.barplot(x='position', y='goals',data=df[df.position != 'Goalkeeper'], estimator=sum)
plt.title('Total goals by position')
plt.show()
```

### 1.4.2.1 Parametro hue
 
Como ya vimos anteriormente, el parámetro hue es un parámetro que encontramos en las funciones de la librería seaborn. En este caso nos permitirá agregar más información al gráfico. Recordad que hue sirve para añadir más información de otra columna categórica. Si ponemos hue='position' veremos no solo los goles por equipo si no por las posiciones de cada equipo.


```
sns.barplot(x='team', y='goals', hue='position', data=df[df.position != 'Goalkeeper'],      
            estimator=sum, ci=False)
plt.title('Goals by positions and teams')
plt.xticks(rotation=90)
plt.show()
```

Ahora podemos ver todavía más información en un solo gráfico. Estamos viendo la cantidad de goles de los diferentes equipos y por posición. Si nos fijamos hay equipos como el Eibar que sus mediocentros marcan más goles que sus delanteros. También vemos como los delanteros del Barcelona marcan muchos más goles que el resto de delanteros.
 
También la propia función sns.barplot() nos crea una leyenda donde vemos a qué categoría corresponde cada color.
 
### 1.4.2.2 Graficos de barras horizontal
 
Para convertir el gráfico de vertical a horizontal, solo tenemos que cambiar el valor de los parámetros x e y. Por tanto los valores numéricos pasarán al eje x y los valores categóricos al eje y.

```
plt.figure(figsize=(12,8))
sns.barplot(x='goals', y='team', hue='position', data=df[df.position !='Goalkeeper'], 
            estimator=sum, ci=False)
plt.title('Total goals by team and position')
plt.show()
```

## 1.4.3.  Plotly
Con la librería plotly podremos crear un gráfico de barras con el método px.bar(). Los parámetros de esta función nos permiten obtener diferentes variantes del gráfico.

```
px.bar(df, x='team', y='goals', title='Goals per team')
```

Si nos fijamos se ven rayas en las barras, si pasamos el ratón por encima, veremos que esta separación entre rayas equivale a los goles de cada jugador en cada equipo. 
 
### 1.4.3.1. Añadir texto
 
Con el parámetro text podremos añadir el nombre de los jugadores pasandole el nombre de la columna que contiene el nombre de los jugadores. Cuando pasemos el ratón por encima podremos saber a quién corresponde esa cantidad de goles.


```
px.bar(df, x='team', y='goals', title='Goals per team', text='player_name')
```

Vemos cómo px.bar() nos permite crear gráficos más interactivos.

### 1.4.3.2. Añadir color por categorías
 
Si añadimos el parámetro color en función de la posición con la columna position, veremos la participación de gol de cada posición



```
px.bar(df, x='team', y='goals', color='position',title='Goals per team')
```

### 1.4.3.3. Parametro barmode
Hemos visto como solo sale una barra con diferentes colores, si queremos que nos salga una barra de cada color por separado, necesitaremos utilizar el parámetro barmode='group'

```
px.bar(df, x='team', y='goals', color='position', barmode='group',title='Goals per team')
```

### 1.4.3.4. Parámetro orientation 
 
Con el parámetro orientation='h' podemos invertir el valor de los ejes, para que se vea de manera horizontal. Recordad siempre intercambiar el valor de los parámetros x e y.  La columnas numérica al eje x y la columna categórica pasa a ser el eje y.


```
px.bar(df, x='goals', y='team', color='position', barmode='group',
       title='Goals per team', orientation='h')
```

### 1.4.3.5. Parámetro facet_row
 
Otra forma de mostrar la misma información, pero de una manera en la que podamos comparar bien las diferentes posiciones de cada equipo, en función del gol, es con el parámetro facet_row. Que nos permite realizar varios gráficos de barras en fila, una por cada valor categórico que contenga la columna que le especifiquemos. En este caso 'position'.


```
px.bar(df[df.position != 'Goalkeeper'], x='team', y='goals',facet_row='position')
```

Ahora podemos comparar visualmente mejor las diferentes posiciones de cada equipo cara al gol. 

#1.5. Pie chart 
 
Otro tipo de gráfico que nos permite graficar una columna categórica y otra numérica es el pie chart, conocido como el gráfico de pastel. Es ideal para mostrar porcentajes en función de una categoría.


## 1.5.1 Matplotlib
Imaginaros que queremos saber el porcentaje de pases que realiza cada posición. Antes de poder crear un gráfico de pastel con matplotlib tendremos que realizar un .groupby() sobre las posiciones y la columna que queramos analizar, en este caso 'passes_total'


```
group = df.groupby('position')['passes_total'].sum()
group
```

Una vez realizada la agrupación pasaremos a crear un gráfico de pastel con las siguientes instrucciones.

```
plt.figure(figsize=(12,8))
labels = ['Forward', 'Defender', 'Midfielder', 'Goalkeeper']
plt.pie(x=group.values, labels=labels, autopct='%1.1f%%', shadow=True)
plt.legend(labels)
plt.show()
```

Pasemos a ver los diferentes parámetros de plt.pie y que hacen: 
 
- x: Le pasamos los valores que queremos graficar, en este caso el valor de nuestra agrupación anterior
- labels: Le especificamos las diferentes porciones, en este caso las diferentes posiciones
- autopct: Con ``'%1.1f%%'`` es una manera de indicar que queremos solo un decimal en los valores numéricos de la proporción del pastel 
- shadow: Añadimos sombra al gráfico. 
 
Con plt.legend() pasándole por parámetro los valores de labels , podremos añadir una leyenda para saber a qué posición corresponde cada color.
 
Como veis nosotros solo realizamos la agrupación extrayendo el total de pases que realiza cada posición. Al pasar esos valores a plt.pie()  nos muestra el porcentaje de pases de cada posición. Podemos ver como el portero y el defensa se llevan la mayoría de pases. 
 
## 1.5.2. Plotly
Con la librería plotly es todavía más sencillo crear un pie chart. No necesitaremos crear una agrupación previa, sólo necesitaremos usa el método px.pie()


```
px.pie(df, values='passes_total', names='position',title='Passes by position')
```

# 1.6. Boxplot
 
En el  boxplot es como si visualizamos las características estadísticas que vimos con la función .describe(). Podemos ver visualmente como se distribuyen los valores de una variable numérica. 
 
En una sola imagen podemos ver: 
 
- mínimo 
- máximo
- mediana
- percentiles 
- outliers


## 1.6.1 Matplotlib
 
Veamos primero un ejemplo realizando un boxplot, de una columna numérica como los tiros a portería, pero primero realizaremos un .describe() a la columna, para ver sus características estadísticas. Esta vez vamos a filtrar por los valores mayores de 0.


```
df[df.shots_total > 0].shots_total.describe()
```

Ahora grafiquemos con plt.boxplot() donde veremos de manera  visual los resultados de  .describe(). Recordad que filtramos los datos para valores superiores a 0.

```
plt.boxplot(df[df.shots_total > 0].shots_total, )
plt.title('Boxplot total shots')
plt.ylabel('Total shots')
plt.show()
```

Esta es la imagen que tiene un boxplot, también conocido como gráfico de caja y bigotes. Realmente nos está mostrando toda la información estadística que nos aporta .describe(). Vayamos por partes y veamos qué información nos aporta las diferentes partes del boxplot.
 
**La caja** 
 
La parte inferior de la caja empieza en el primer cuartil o percentil 25% y termina en el tercer cuartil o percentil 75%. La raya que corta por la mitad la caja, hace referencia al percentil 50% que también es la mediana. Por tanto dentro de la caja se acumulan el 50% de los datos, ya que 75-25 = 50. De esta manera vemos que valores ronda el 50% de los datos de la columna shots_total. Estos tipos de gráficos nos enseñan a entender cómo se distribuyen los datos. 
 
**Las líneas o bigotes inferior y superior de la caja.**
 
Las rayas que salen de la caja que se les conoce como los bigotes, se calculan de la siguiente manera:

v9.png


Como vemos la raya inferior se obtiene de restar 1.5 x IQR al primer cuartil que es percentil 25% . El IQR es el rango intercuartílico que es el tamaño de la caja y que obtenemos restando el tercer cuartil o percentil 75% al primer cuartil o percentil 25%. Todo los datos que estén fuera de estas rayas se les conoce como valores Outliers. Un outlier es un valor el cual se encuentra demasiado alejado al resto de valores. Nuevamente en los tiros a puerta vemos la característica de que muchos jugadores tiran poco, pero unos pocos tiran mucho. 
 
A veces los valores mínimos y máximos serán el final de los bigotes, pero en caso de que tengamos outliers, serán estos los valores máximos y mínimos de nuestra columna.
